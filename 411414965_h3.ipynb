```python
days, hours = divmod(384400 // 1225, 24)
result = f"共需：{days}天又{hours}小時"
with open("text.test", "w", encoding="utf-8") as f:
    f.write("計算公式：總小時 = 384400 // 1225\n")
    f.write(f"輸出結果：{result}")
```
